```python
from pathlib import Path
import numpy as np
import pandas as pd
import json
import hashlib
from datetime import datetime

project_dir = Path(
    r"C:\Users\praja\functional-connectivity\neuroscience_project"
)

results_dir = project_dir / "results"

results_dir.mkdir(exist_ok=True)

print("Project:", project_dir)
print("Results:", results_dir)
```

```python
required_files = [
    "distance_matrix.npy",
    "persistence_H0.npy",
    "persistence_H1.npy",
    "milestone8_threshold_sensitivity.csv",
    "milestone9_network_metrics.csv",
    "milestone9_persistence_summary.csv"
]

for filename in required_files:

    path = results_dir / filename

    print(
        filename,
        "FOUND" if path.exists() else "MISSING"
    )

missing = [
    name for name in required_files
    if not (results_dir / name).exists()
]

if missing:
    raise FileNotFoundError(
        f"Missing project files: {missing}"
    )

print("\nPROJECT FILES VERIFIED")
```
